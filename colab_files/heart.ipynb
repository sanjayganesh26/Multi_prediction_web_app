```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score
import pickle
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
heart=pd.read_csv('/content/heart.csv')
heart.head()
```

```python
heart.isnull().sum()
```

```python
heart['target'].value_counts()
```

```python
X=heart.drop(columns='target',axis=1)
Y=heart['target']
```

```python
models = [LogisticRegression(max_iter=10000),SVC(kernel='linear',C=5),KNeighborsClassifier(),RandomForestClassifier(n_estimators=100)]

def CV_score():
  for model in models:
    cv_score=cross_val_score(model,X,Y,cv=5)
    mean_accuracy=sum(cv_score)/len(cv_score)
    mean_accuracy=mean_accuracy*100
    mean_accuracy=round(mean_accuracy,2)
    print(model , ':' , mean_accuracy , '%')
    print('-----------------------------------------------')
```

```python
CV_score()
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=3)
print(X.shape , x_train.shape , x_test.shape)
```

```python
model1=LogisticRegression(max_iter=10000)
model1.fit(x_train,y_train)
```

```python
x_train_predict=model1.predict(x_train)
training_data_accuracy=accuracy_score(y_train,x_train_predict)
print(training_data_accuracy)
```

```python
x_test_predict=model1.predict(x_test)
test_data_accuracy=accuracy_score(y_test,x_test_predict)
print(test_data_accuracy)
```

```python
filename = 'heart_model.sav'
pickle.dump(model1, open(filename, 'wb'))
```
